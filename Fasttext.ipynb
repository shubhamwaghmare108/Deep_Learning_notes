```python
from gensim.models import FastText

sentences = [
    ["i", "love", "playing"],
    ["he", "is", "running"],
    ["they", "play", "football"]
]

# Train FastText model
model = FastText(sentences, vector_size=10, window=3, min_count=1)

# Get word vector
print(model.wv["playing"])

# Similar words
print(model.wv.most_similar("playing"))
```
