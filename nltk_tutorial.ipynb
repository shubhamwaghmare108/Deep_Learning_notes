```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet

text = "The cats are running and playing"

tokens = word_tokenize(text)

pos_tags = nltk.pos_tag(tokens)

print(pos_tags)

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):

    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

lemmas = []

for word, tag in pos_tags:
    
    wn_tag = get_wordnet_pos(tag)
    
    lemma = lemmatizer.lemmatize(word, pos=wn_tag)
    
    lemmas.append(lemma)

print(lemmas)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "I am learning natural language processing"

tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
"I love NLP",
"I love machine learning",
"NLP is amazing"
]

vectorizer = CountVectorizer(ngram_range=(1,3)) # (1,1) for unigrams, (2,2) for bigrams, (1,2) for both

X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
from nltk.probability import FreqDist
from nltk.tokenize import word_tokenize

text = "data science is fun and data science is powerful"

tokens = word_tokenize(text)

fdist = FreqDist(tokens)

print(fdist.most_common(5))
```

```python
from gensim.models import Word2Vec

# Sample corpus
sentences = [
    ["i", "love", "data"],
    ["data", "is", "powerful"],
    ["i", "love", "machine", "learning"]
]

# Train model
model = Word2Vec(sentences, vector_size=100, window=3, min_count=1)

# Get vector
print(model.wv["data"])

# Similar words
print(model.wv.most_similar("data"))
```
