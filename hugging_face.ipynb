```python
from transformers import pipeline

```

```python
classifier = pipeline("sentiment-analysis")

result = classifier("I love learning NLP")

print(result)
```

```python


generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

result = generator("Artificial intelligence is", max_length=30)

print(result)
```

```python
from transformers import pipeline

ner = pipeline("ner")

text = "Elon Musk founded SpaceX"

result = ner(text)

print(result)
```
