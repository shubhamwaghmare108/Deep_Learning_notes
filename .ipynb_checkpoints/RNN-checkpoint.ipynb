```python

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
# sample data
X = np.array([
    [[1],[2],[3]],
    [[2],[3],[4]],
    [[3],[4],[5]]
])

y = np.array([4,5,6])
```

```python
# model
model = Sequential()

model.add(SimpleRNN(10, activation='tanh', input_shape=(3,1)))

model.add(Dense(1))

model.compile(
    optimizer='adam',
    loss='mse'
)

model.fit(X,y,epochs=100)
```

```python
print(model.predict(np.array([[[4],[5],[6]]])))
```
