```python
import nltk

# Tokenizer models (required for word_tokenize and sent_tokenize)
nltk.download('punkt')

# Newer versions of NLTK sometimes require this table for punkt tokenizer
nltk.download('punkt_tab')

# Stopword list (used to remove common words like "the", "is", "in")
nltk.download('stopwords')

# WordNet lexical database (used for lemmatization)
nltk.download('wordnet')

# WordNet multilingual dictionary mapping
nltk.download('omw-1.4')

# Part-of-Speech tagging model
# Used for identifying nouns, verbs, adjectives etc.
nltk.download('averaged_perceptron_tagger')

# Named Entity Recognition chunker model
# Used for detecting entities like PERSON, LOCATION, ORGANIZATION
nltk.download('maxent_ne_chunker')

# Word list used by the NER chunker
nltk.download('words')

# VADER sentiment lexicon
# Used for sentiment analysis (positive, negative, neutral)
nltk.download('vader_lexicon')

```

```python
# Movie review dataset (used for sentiment analysis practice)
nltk.download('movie_reviews')

# Twitter sample dataset (used for NLP experiments)
nltk.download('twitter_samples')

# Brown corpus (one of the oldest NLP datasets)
nltk.download('brown')

# Gutenberg dataset (collection of books)
nltk.download('gutenberg')
```

```python
documents = []
labels = []

for category in movie_reviews.categories():
    
    for fileid in movie_reviews.fileids(category):
        
        text = movie_reviews.raw(fileid)
        
        documents.append(text)
        labels.append(category)

print(len(documents))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    documents,
    labels,
    test_size=0.2,
    random_state=42
)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

def clean_text(text):
    
    text = text.lower()
    
    text = re.sub(r'[^\w\s]', '', text)
    
    tokens = word_tokenize(text)
    
    tokens = [w for w in tokens if w not in stop_words]
    
    return " ".join(tokens)
```

```python
X_train = [clean_text(t) for t in X_train]
X_test = [clean_text(t) for t in X_test]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(max_features=5000)

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X_train_vec, y_train)
```

```python
from sklearn.metrics import accuracy_score

pred = model.predict(X_test_vec)

print("Accuracy:", accuracy_score(y_test, pred))
```

```python
test_review = "This movie was fantastic and the acting was great"

clean = clean_text(test_review)

vec = vectorizer.transform([clean])

prediction = model.predict(vec)

print(prediction)
```

```python
test_review = "This movie was terrible and the acting was bad"

clean = clean_text(test_review)

vec = vectorizer.transform([clean])

prediction = model.predict(vec)

print(prediction)
```

```python
def predict_sentiment(review):
    
    clean = clean_text(review)
    
    vec = vectorizer.transform([clean])
    
    prediction = model.predict(vec)
    
    return prediction[0]    
```

```python
prediction = predict_sentiment("I really enjoyed this movie, it was fantastic!")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(max_features=5000)

X_train_vec = vectorizer.fit_transform(X_train)

X_test_vec = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model1 = LogisticRegression()

model1.fit(X_train_vec, y_train)
```

```python
from sklearn.metrics import accuracy_score

pred1 = model1.predict(X_test_vec)

print("Accuracy:", accuracy_score(y_test, pred1))
```

```python
def predict_sentiment(review):
    
    clean = clean_text(review)
    
    vec = vectorizer.transform([clean])
    
    prediction = model1.predict(vec)
    
    return prediction[0]    
```
