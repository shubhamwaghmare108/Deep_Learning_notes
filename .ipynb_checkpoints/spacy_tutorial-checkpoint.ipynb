```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("Apple is looking at buying a UK startup for $1 billion")
```

```python
for token in doc:
    print(token.text)
```

```python
for token in doc:
    print(token.text, token.pos_, token.lemma_)
```

```python
for token in doc:
    print(token.text, token.is_stop)
```

```python
for token in doc:
    print(token.text, "→", token.lemma_)
```

```python
for token in doc:
    print(token.text, token.pos_)
```

| POS  | Meaning   |
| ---- | --------- |
| NOUN | noun      |
| VERB | verb      |
| ADJ  | adjective |
| ADV  | adverb    |


```python
doc = nlp("Apple is buying a startup in London for $1 billion")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
doc = nlp("shubham is a software engineer at Microsoft")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
doc = nlp("I love NLP. It is amazing.")

for sent in doc.sents:
    print(sent)
```

```python
doc1 = nlp("dog")
doc2 = nlp("cat")

print(doc1.similarity(doc2))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "I am learning Natural Language Processing"

doc = nlp(text)

tokens = [
    token.lemma_.lower()
    for token in doc
    if not token.is_stop and not token.is_punct
]

print(tokens)
```

```python
nlp.Defaults.stop_words.add("nlp")
```

```python
print("nlp" in nlp.Defaults.stop_words)
```

```python
print(nlp.pipe_names)
```

```python
texts = [
"I love NLP",
"Machine learning is powerful",
"spaCy is fast"
]

for doc in nlp.pipe(texts):
    print([token.text for token in doc])
```
