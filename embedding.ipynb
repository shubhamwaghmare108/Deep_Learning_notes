```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["i","love","nlp"],
    ["nlp","is","fun"],
    ["machine","learn","is","powerful"],
    ["i","love","machine","learning"]
]
```

```python
model = Word2Vec(sentences, vector_size=15, window=2, min_count=1)
```

```python
print(model.wv["learning"])
```

```python
print(model.wv["learn"])
```

```python
print(model.wv['love'])
```

```python
print(model.wv.most_similar('learn'))
```

```python
model.wv.most_similar('fun')
```

```python
from transformers import BertTokenizer, BertModel
import torch

# load tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
sentence1 = "I deposited money in the bank"
sentence2 ="I visited ganga river bank"
```

```python
inputs1 = tokenizer(sentence1, return_tensors="pt")
inputs2 = tokenizer(sentence2, return_tensors="pt")
```

```python
with torch.no_grad():
    outputs1 = model(**inputs1)
    outputs2 = model(**inputs2)
```

```python
tokens1 = tokenizer.convert_ids_to_tokens(inputs1['input_ids'][0])
tokens2 = tokenizer.convert_ids_to_tokens(inputs2['input_ids'][0])

bank_index1 = tokens1.index("bank")
bank_index2 = tokens2.index("bank")
```

```python
embedding1 = outputs1.last_hidden_state[0][bank_index1]
embedding2 = outputs2.last_hidden_state[0][bank_index2]
```

```python
from torch.nn.functional import cosine_similarity

similarity = cosine_similarity(embedding1, embedding2, dim=0)

print("Similarity:", similarity.item())
```
